# Vehicle Engine RPM Forecasting using OBD-II Data with Granite Time Series - Few-shot finetuning, evaluation, and visualization

In this tutorial, we will explore [timeseries forecasting](https://www.ibm.com/think/insights/time-series-forecasting) using the [IBM Granite Timeseries model](https://ibm.com/granite) to predict vehicle Engine RPM. We will cover key techniques such as few-shot forecasting and fine-tuning. We are using the [KIT OBD-II dataset](https://publikationen.bibliothek.kit.edu/1000078727) which contains real-world vehicle sensor data collected from a Seat Leon during various driving conditions. The aim of this notebook is to showcase how to use a pre-trained time series foundation model for multivariate forecasting of automotive sensor data and explore various features available with Granite Time Series Foundation Models.

This notebook uses TinyTimeMixers (TTMs), which are compact pre-trained models for Multivariate Time-Series Forecasting, open-sourced by IBM Research. With less than 1 Million parameters, TTM introduces the notion of the first-ever "tiny" pre-trained models for Time-Series Forecasting. TTM outperforms several popular benchmarks demanding billions of parameters in zero-shot and few-shot forecasting and can easily be fine-tuned for multivariate forecasts.

### Specify configuration variables

The forecast length is specified as well as the context length (in time steps) which is set to match the pretrained model. Additionally, we declare the Granite Time Series Model and the specific revision we are targeting.

The granite-timeseries TTM R2 card has several different revisions of the model available for various context lengths and prediction lengths. In this example we will be working with daily data, so we choose a model suitable for that resolution -- 90 days of history to forecast the next 30 days.

```python
forecast_length = 28
context_length = 90

TTM_MODEL_PATH = "ibm-granite/granite-timeseries-ttm-r2"
REVISION = "90-30-ft-l1-r2.1"

# Use MPS (Metal Performance Shaders) for Apple Silicon GPU acceleration
if torch.backends.mps.is_available():
    device = "mps"
    print("✓ Using Apple Silicon GPU (MPS)")
elif torch.cuda.is_available():
    device = "cuda"
    print("✓ Using CUDA GPU")
else:
    device = "cpu"
    print("⚠ Using CPU (slower)")
```

## Preparing the Data

As mentioned in the introduction, this notebook makes use of the [KIT OBD-II dataset](https://publikationen.bibliothek.kit.edu/1000078727).

The original data contains real-world vehicle sensor measurements collected from a Seat Leon during various driving conditions including normal traffic, free-flowing traffic, and congestion (Stau). The data includes multiple OBD-II parameters such as Engine RPM, Vehicle Speed, Coolant Temperature, Throttle Position, and various other sensor readings. Each CSV file represents a different driving session with specific routes and traffic conditions. The combined dataset aggregates these sessions with metadata about source files and traffic conditions for our forecasting task.


### Read in the data

We load the combined OBD-II dataset into a pandas dataframe. The dataset includes a timestamp column, various sensor measurements (Engine RPM, Vehicle Speed, Coolant Temperature, etc.), and metadata fields (source_file, traffic_condition). We drop unnecessary columns that were used during data preprocessing but aren't needed for forecasting.

```python
dataset_path = "/Users/darenpalmer/Desktop/UCL/CS/fyp.nosync/data/combined_obd_dataset_watsonx.csv"
print(f"\nLoading combined dataset from: {dataset_path}")
data = pd.read_csv(dataset_path, parse_dates=["timestamp"])
```

```python
data = data.drop(columns=["vehicle", "Time_original", "unique_id", "date", "original_row_id", "route_from", "route_to"])
print(f"✓ Loaded: {len(data):,} total samples")
print(f"✓ Columns: {len(data.columns)}")

# Preview
print(f"\nDataset preview:")
data.head()
```

Next, we must configure the columns in our data and declare the names of the timestamp column, the target column to be predicted (Engine RPM), the ID column used to group time series (source_file), control columns (exogenous features like Vehicle Speed, Coolant Temperature, etc.), and the static categorical column (traffic_condition) that describes the driving conditions for each session.

```python
cols = list(data.columns)
[cols.remove(c) for c in ["timestamp", "Engine RPM [RPM]", "source_file", "traffic_condition"]]
cols

column_specifiers = {
    "timestamp_column": "timestamp",
    "id_columns": ["source_file"],
    "target_columns": ["Engine RPM [RPM]"],
    "control_columns": cols,
    "static_categorical_columns": ["traffic_condition"],
}
```

### Train the Preprocessor

The preprocessor is trained on the input data to learn the scaling factors for normalization. We split the data by source files (driving sessions) rather than by time to ensure each session is either in training or testing, preventing data leakage. The scaling will be applied when we use the preprocess method of the time series preprocessor. Note that some warnings may appear due to constant or near-constant values in certain sensor readings during idle periods.

```python
tsp = TimeSeriesPreprocessor(
    **column_specifiers,
    context_length=context_length,
    prediction_length=forecast_length,
    scaling=True,
    encode_categorical=True,
    scaler_type="standard",
)

# Group data by source file
grouped_data = data.groupby(column_specifiers["id_columns"])

# Get the list of unique source files
source_files = list(grouped_data.groups.keys())

# Determine the split point (e.g., 80% for training, 20% for testing)
split_index = int(len(source_files) * 0.8)

# Split source files into training and testing sets
train_files = source_files[:split_index]
test_files = source_files[split_index:]

# Create training and testing dataframes
df_train = data[data[column_specifiers["id_columns"][0]].isin(train_files)].copy()
df_test = data[data[column_specifiers["id_columns"][0]].isin(test_files)].copy()

# Train the preprocessor on the entire dataset
trained_tsp = tsp.train(data)
```

## Finetune the model

Now we will focus on fine-tuning the pretrained model. We use the data splits defined above (by source file), and include the exogenous control columns (Vehicle Speed, Coolant Temperature, etc.) and static categorical features (traffic_condition) during the fine-tuning process.

### Preparing the data for fine-tuning

We split the preprocessed data into training, validation, and test sets. The training files (80% of source files) are further split into training (75%) and validation (25%) portions. The test set (20% of source files) is used to evaluate final performance.

```python
# Preprocess the entire dataset
data_preprocessed = tsp.preprocess(data)

split_params = {"train": 0.75, "test": 0.25} # Adjusted split ratios for within train_data_preprocessed

# Create training and testing dataframes from the preprocessed data
train_data_preprocessed = data_preprocessed[data_preprocessed[column_specifiers["id_columns"][0]].isin(train_files)].copy()
test_data_preprocessed = data_preprocessed[data_preprocessed[column_specifiers["id_columns"][0]].isin(test_files)].copy()

# Split the preprocessed training data into training and validation sets
train_data, valid_data, _ = prepare_data_splits(
    train_data_preprocessed, id_columns=column_specifiers["id_columns"], split_config=split_params, context_length=context_length
)

test_data = test_data_preprocessed # Use the preprocessed df_test as the test set
```

Here we construct PyTorch datasets from the pandas dataframes using the ForecastDFDataset class, which is specifically designed for forecasting use cases. Note: You may see a warning about the frequency token not being found in the mapping, which is expected for high-frequency OBD-II data (~90ms intervals) that differs from typical time series frequencies.

```python
frequency_token = tsp.get_frequency_token(tsp.freq)

dataset_params = column_specifiers.copy()
dataset_params["frequency_token"] = frequency_token
dataset_params["context_length"] = context_length
dataset_params["prediction_length"] = forecast_length


train_dataset = ForecastDFDataset(train_data, **dataset_params)
valid_dataset = ForecastDFDataset(valid_data, **dataset_params)
test_dataset = ForecastDFDataset(test_data, **dataset_params)
```

Now let's take a smaller sample from the torch datasets produced above for few-shot fine-tuning. We use 20% of the training and validation data, which is a common approach when working with foundation models that have been pretrained on large amounts of data.

```python
# 20% training and validation data (few-shot finetuning)
fewshot_fraction = 0.20
n_train_all = len(train_dataset)
train_index = np.random.permutation(n_train_all)[: int(fewshot_fraction * n_train_all)]
train_dataset = Subset(train_dataset, train_index)

n_valid_all = len(valid_dataset)
valid_index = np.random.permutation(n_valid_all)[: int(fewshot_fraction * n_valid_all)]
valid_dataset = Subset(valid_dataset, valid_index)

n_train_all, len(train_dataset), n_valid_all, len(valid_dataset)
```

### Load the model for fine-tuning

We load the TTM (TinyTimeMixer) model from HuggingFace using the model path and revision configured above. Our OBD-II data has one target channel (Engine RPM), several exogenous channels (Vehicle Speed, Coolant Temperature, Throttle Position, etc.), and one static categorical input (traffic_condition). The `TimeSeriesPreprocessor` provides the `prediction_channel_indices`, `exogenous_channel_indices`, and `categorical_vocab_size_list` information to the model. We enable channel mixing in the decoder (`decoder_mode="mix_channel"`) and forecast channel mixing (`enable_forecast_channel_mixing=True`) to capture interactions between the sensor channels and adjust forecasts based on exogenous variables. The `fcm_context_length` parameter controls how much historical lag from exogenous variables influences the forecast.

```python
set_seed(1234)

finetune_forecast_model = TinyTimeMixerForPrediction.from_pretrained(
    TTM_MODEL_PATH,
    revision=REVISION,
    context_length=context_length,
    prediction_filter_length=forecast_length,
    num_input_channels=tsp.num_input_channels,
    decoder_mode="mix_channel",  # exog:  set to mix_channel for mixing channels in history
    prediction_channel_indices=tsp.prediction_channel_indices,
    exogenous_channel_indices=tsp.exogenous_channel_indices,
    fcm_context_length=1,  # exog: indicates lag length to use in the exog fusion. for Ex. if today sales can get affected by discount on +/- 2 days, mention 2
    fcm_use_mixer=True,  # exog: Try true (1st option) or false
    fcm_mix_layers=2,  # exog: Number of layers for exog mixing
    enable_forecast_channel_mixing=True,  # exog: set true for exog mixing
    categorical_vocab_size_list=tsp.categorical_vocab_size_list,  # sizes of the static categorical variables
    fcm_prepend_past=True,  # exog: set true to include lag from history during exog infusion.
)
```

### Optional: Freeze the TTM Backbone

Oftentimes, during fine-tuning we freeze the backbone and focus on tuning only the parameters in the decoder. This reduces the overall number of parameters being tuned and maintains what the encoder learned during pretraining.

For this OBD-II dataset, we found that performance was better when the backbone remained unfrozen, likely because the high-frequency automotive sensor data has different characteristics than the data used for pretraining. For other datasets, freezing the backbone might be preferable. We have disabled the backbone freezing code but left it intact as an example.

```python
freeze_backbone = False
if freeze_backbone:
    print(
        "Number of params before freezing backbone",
        count_parameters(finetune_forecast_model),
    )

    # Freeze the backbone of the model
    for param in finetune_forecast_model.backbone.parameters():
        param.requires_grad = False

    # Count params
    print(
        "Number of params after freezing the backbone",
        count_parameters(finetune_forecast_model),
    )
```

### Set up a Trainer for Fine-tuning

Configure a Trainer for fine-tuning and evaluating the model. We use the learning rate finder to automatically identify an optimal learning rate for our OBD-II dataset.

```python
num_epochs = 50
# Optimized batch size for Apple Silicon M2 with 16GB unified memory
# Adjust if you have different hardware capabilities
batch_size = 32

learning_rate, finetune_forecast_model = optimal_lr_finder(
    finetune_forecast_model,
    train_dataset,
    batch_size=batch_size,
    enable_prefix_tuning=True,
)
print("OPTIMAL SUGGESTED LEARNING RATE =", learning_rate)
```

### Train the Model

Here we train the model on the OBD-II training data using the Hugging Face Trainer. We use early stopping to prevent overfitting and the OneCycleLR scheduler for better convergence. The model is trained on Apple Silicon GPU (MPS) or CUDA if available.

```python
OUT_DIR = "ttm_finetuned_models/"

print(f"Using learning rate = {learning_rate}")
print(f"Using device = {device}")

finetune_forecast_args = TrainingArguments(
    output_dir=os.path.join(OUT_DIR, "output"),
    overwrite_output_dir=True,
    learning_rate=learning_rate,
    num_train_epochs=num_epochs,
    do_eval=True,
    eval_strategy="epoch",
    per_device_train_batch_size=batch_size,
    per_device_eval_batch_size=2 * batch_size,
    dataloader_num_workers=2,  # Increased for better data loading on M2
    report_to="none",
    save_strategy="epoch",
    logging_strategy="epoch",
    save_total_limit=1,
    logging_dir=os.path.join(OUT_DIR, "logs"),
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    use_cpu=False,  # Always use accelerator (MPS or CUDA) when available
    fp16=False,  # MPS doesn't support fp16 yet
    # Optimization: gradient accumulation for effective larger batch size
    gradient_accumulation_steps=2,  # Effective batch size = 32 * 2 = 64
)

# Create the early stopping callback
early_stopping_callback = EarlyStoppingCallback(
    early_stopping_patience=5,  # Number of epochs with no improvement after which to stop
    early_stopping_threshold=0.001,  # Minimum improvement required to consider as improvement
)
tracking_callback = TrackingCallback()

# Optimizer and scheduler
optimizer = AdamW(finetune_forecast_model.parameters(), lr=learning_rate)
scheduler = OneCycleLR(
    optimizer,
    learning_rate,
    epochs=num_epochs,
    steps_per_epoch=math.ceil(len(train_dataset) / (batch_size)),
)

finetune_forecast_trainer = Trainer(
    model=finetune_forecast_model,
    args=finetune_forecast_args,
    train_dataset=train_dataset,
    eval_dataset=valid_dataset,
    callbacks=[early_stopping_callback, tracking_callback],
    optimizers=(optimizer, scheduler),
)

# Fine tune
finetune_forecast_trainer.train()

finetune_forecast_trainer.evaluate(test_dataset)
```

### Evaluate the Model

Evaluate the fine-tuned model on the test dataset using standard regression metrics (MSE, RMSE, MAE) to assess the accuracy of Engine RPM predictions.

```python
# Define standard regression metrics for evaluating forecasts
def custom_metric(actual, prediction, column_header="results"):
    """
    Compute MSE, RMSE, and MAE for forecast evaluation.
    
    Args:
        actual: Actual values (pandas Series or array-like)
        prediction: Predicted values (pandas Series or array-like)
        column_header: Header for the results column in output DataFrame
    
    Returns:
        DataFrame with computed metrics
    """
    a = np.asarray(actual.tolist())
    p = np.asarray(prediction.tolist())
    if p.shape[1] < a.shape[1]:
        a = a[:, : p.shape[1]]

    # Create mask to exclude NaN values
    mask = ~np.any(np.isnan(a), axis=1)

    mse = np.mean(np.square(a[mask, :] - p[mask, :]))
    mae = np.mean(np.abs(a[mask, :] - p[mask, :]))
    return pd.DataFrame(
        {
            column_header: {
                "mean_squared_error": mse,
                "root_mean_squared_error": np.sqrt(mse),
                "mean_absolute_error": mae,
            }
        }
    )
```

```python
# Generate Engine RPM forecasts using the fine-tuned model
pipeline = TimeSeriesForecastingPipeline(
    finetune_forecast_model,
    device=device,  # Use MPS (Apple Silicon GPU) or CUDA
    feature_extractor=tsp,
    batch_size=batch_size,
)

# Make forecasts on the test data
finetune_forecast = pipeline(test_data)
finetune_forecast.head()
```

```python
# Compute metrics for Engine RPM forecasts
custom_metric(
    finetune_forecast["Engine RPM [RPM]"], 
    finetune_forecast["Engine RPM [RPM]_prediction"], 
    "fine-tune forecast"
)
```
